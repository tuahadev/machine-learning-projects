```python
df=pd.read_csv('fake_or_real_news.csv')
df.head()
```

```python
df.isna().sum()
```

```python
x=df['text'].astype(str).values
y=df['label'].map({'FAKE':0,'REAL':1})
y
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
tfid=TfidfVectorizer(max_features=5000,stop_words='english')
X_train_tfidf = tfid.fit_transform(x_train).toarray()
X_test_tfidf = tfid.transform(x_test).toarray()
```

```python
model = models.Sequential([
    layers.Dense(128, activation='relu', input_shape=(X_train_tfidf.shape[1],)),
    layers.Dropout(0.3),
    layers.Dense(64, activation='relu'),
    layers.Dropout(0.2),
    layers.Dense(1, activation='sigmoid')
])
model.summary()
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])

model.summary()

model.fit(X_train_tfidf, y_train, epochs=10, batch_size=32)
```

```python
model.evaluate(X_test_tfidf,y_test)
y_pred=model.predict(X_test_tfidf)
```

```python
# Correct prediction code:
user_input = input("Enter news article text: ")
user_idf=tfid.transform([user_input]).toarray()
prediction = model.predict(user_idf)
print("Prediction:", "REAL" if prediction[0] > 0.5 else "FAKE")
```

```python

y_pred_binary = (y_pred > 0.5).astype(int)

from sklearn.metrics import classification_report, confusion_matrix

print(classification_report(y_test, y_pred_binary, target_names=['REAL', 'FAKE']))


print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_binary))

```
